```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
```

```python
# Créer une nouvelle présentation
prs = Presentation()

# Couleurs inspirées du logo DGI
VERT = RGBColor(6, 74, 31)
ROUGE = RGBColor(206, 17, 38)
JAUNE = RGBColor(252, 209, 22)
TEXTE_NOIR = RGBColor(0, 0, 0)

# Thème par défaut : police Arial
prs.slide_width = Inches(13.33)
prs.slide_height = Inches(7.5)
```

```python
# Fonction pour ajouter une slide avec titre et contenu
def add_slide(prs, title, bullets=[]):
    layout = prs.slide_layouts[1]  # Titre + Contenu
    slide = prs.slides.add_slide(layout)
    
    # Titre
    slide.shapes.title.text = title
    slide.shapes.title.text_frame.paragraphs[0].font.size = Pt(40)
    slide.shapes.title.text_frame.paragraphs[0].font.bold = True
    slide.shapes.title.text_frame.paragraphs[0].font.color.rgb = VERT
    
    # Contenu
    body = slide.placeholders[1]
    tf = body.text_frame
    tf.clear()
    for b in bullets:
        p = tf.add_paragraph()
        p.text = b
        p.font.size = Pt(28)
        p.font.color.rgb = TEXTE_NOIR
        p.level = 0
```

```python
# Exemple d'utilisation
add_slide(prs, "Titre de la présentation", ["Point 1", "Point 2", "Point 3"])
add_slide(prs, "Introduction", ["Contexte", "Objectifs", "Méthodologie"])
add_slide(prs, "Résultats", ["Résultat 1", "Résultat 2", "Graphiques"])

# Ajouter un pied de page avec logo (optionnel)
# slide.shapes.add_picture("logo_dgi.png", Inches(11), Inches(6.8), width=Inches(1.5))

# Enregistrer le fichier PowerPoint
prs.save("Memoire_DGI_Theme.pptx")

print("✅ PowerPoint généré avec succès : Memoire_DGI_Theme.pptx")
```
